# IMDb Sentiment: Multinomial Naive Bayes

The same 50,000 IMDb reviews, this time with a Bag-of-Words representation (CountVectorizer, 10,000 features) and a Multinomial Naive Bayes classifier. The trained model and vectorizer are saved with `pickle` for reuse.

**Data:** `IMDB Dataset.csv` (see [`../data/README.md`](../data/README.md)).

```python
# Import the libraries
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import pickle

DATA_DIR = "../data/"
```

```python
# Load the dataset (`IMDB Dataset.csv`)
df = pd.read_csv(DATA_DIR + 'IMDB Dataset.csv', encoding='latin-1', header=0)
```

```python
df
```

```python
# Convert the labels to numbers
df['sentiment'] = df['sentiment'].replace({
    'negative': 0,
    'positive': 1
})
```

```python
# Check the result
print(df['sentiment'].unique())
```

```python
# Number of positive and negative reviews
print(df['sentiment'].value_counts())
```

```python
# Step 2: Text preprocessing with CountVectorizer
# Turn the text into a Bag-of-Words representation
df['review'] = df['review'].astype(str).str.lower()
vectorizer = CountVectorizer(stop_words='english', max_df=0.8, min_df=10, max_features=10000)
X = vectorizer.fit_transform(df['review'])
y = df['sentiment']
```

```python
# Step 3: Split into training and test sets (70/30)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# Step 4: Train the Naive Bayes classifier
modelo = MultinomialNB()
modelo.fit(X_train, y_train)
```

```python
# Step 5: Predict
y_pred = modelo.predict(X_test)

# Step 6: Evaluate the model
print("\nPrecisión del modelo:", accuracy_score(y_test, y_pred))
print("\nMatriz de Confusión:")
print(confusion_matrix(y_test, y_pred))
print("\nReporte de Clasificación:")
print(classification_report(y_test, y_pred))
```

```python
# Step 7: Save the Naive Bayes model and the CountVectorizer
with open('naive_bayes_model.pkl', 'wb') as model_file:
    pickle.dump(modelo, model_file)

with open('count_vectorizer.pkl', 'wb') as vectorizer_file:
    pickle.dump(vectorizer, vectorizer_file)
```

## Results

Naive Bayes reached **84.7% accuracy** on 15,000 test reviews (precision and recall around 0.84–0.86 for both classes), below the 88.4% of Logistic Regression in notebook 01.
